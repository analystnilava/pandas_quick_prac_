# CONCATINATE AND MARGE TEXT

```python
d1 = pd.DataFrame({'ID' : [1 , 2 , 3] , 'NAME' : ['A' , 'B' , 'C']})
d2 = pd.DataFrame({'ID' : [1, 2 , 2, 4] , 'Score' : [800 , 40 , 60 , 50]})
```

```python
pd.concat([d1, d2] , axis=0)
```

```python
pd.concat([d1, d2] , axis=1)
```

#marge(join)

```python
pd.merge(d1 , d2 , how='inner' , on='ID')
```

```python
pd.merge(d1 , d2 , how='inner' , left_on='ID' , right_on='ID')
```
